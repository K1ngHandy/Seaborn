import seaborn

```python
import seaborn as sns
```

apply the default theme

```python
sns.set_theme()
```

load example dataset

```python
tips = sns.load_dataset("tips")
```

Create visualization

```python
sns.relplot(
    data = tips,
    x = "total_bill", y = "tip", col = "time",
    hue = "smoker", style = "smoker", size = "size",
)
```
